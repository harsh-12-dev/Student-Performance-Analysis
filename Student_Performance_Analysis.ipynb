### Data Science & Data Analytics Using Python

This project analyzes student performance data to understand academic performance, study habits, and gender-wise differences in final grades.

### Objectives
- Explore and clean the student performance dataset.
- Calculate the average final grade.
- Count students scoring above 15.
- Analyze the relationship between study time and final grades.
- Compare average performance between male and female students.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv('student-mat.csv', sep=';')

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Check missing values after cleaning
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

## Data Cleaning

The dataset was checked for duplicate records and missing values.

- Duplicate rows were identified and removed.
- Missing values were checked for all columns.
- The dataset contains no missing values after the cleaning process.
- The cleaned dataset is now ready for further analysis.

```python
average_g3 = df['G3'].mean()

print(f"Average Final Grade (G3): {average_g3:.2f}")
```

```python
students_above_15 = (df['G3'] > 15).sum()

print(f"Number of students scoring above 15: {students_above_15}")
```

```python
studytime_g3_correlation = df['studytime'].corr(df['G3'])

print(f"Correlation between Study Time and Final Grade: {studytime_g3_correlation:.3f}")
```

```python
gender_average = df.groupby('sex')['G3'].mean()

print("Average Final Grade by Gender:")
print(gender_average)
```

## Data Visualization

Visualizations help us understand the distribution of final grades and the relationship between study time, academic performance, and gender.

```python
plt.figure(figsize=(8, 5))

plt.hist(df['G3'], bins=11, edgecolor='black')

plt.title('Distribution of Final Grades (G3)')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['studytime'], df['G3'], alpha=0.6)

plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')

plt.show()
```

```python
gender_average = df.groupby('sex')['G3'].mean()

plt.figure(figsize=(7, 5))

plt.bar(
    ['Female', 'Male'],
    [gender_average['F'], gender_average['M']],
    edgecolor='black'
)

plt.title('Average Final Grade by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Final Grade (G3)')

plt.show()
```

### Visualization Insights

- The histogram shows the distribution of students' final grades and helps identify the most common grade range.
- The scatter plot provides a visual representation of the relationship between study time and final grades.
- The bar chart compares the average final grade of female and male students.
